```python
def read_all(ser):
    out = []
    while ser.in_waiting > 0:
        data1b = ser.read(1)
        data1 = data1b.decode('utf-8')
        out.append(data1)
        
    outstr = ''.join(out)
    return outstr
```

```python
read_all(ser)
```

```python
def read_one_line(ser):
    out = []
    while ser.in_waiting > 0:
        data1b = ser.read(1)
        data1 = data1b.decode('utf-8')
        if data1 in ['\n','\r']:
            break
        out.append(data1)
        
    outstr = ''.join(out)
    return outstr
```

```python
read_one_line(ser)
```

```python
read_all(ser)
```

```python
def one_byte_int_to_serial_byte(int_byte):
    out_byte = int_byte.to_bytes(1,byteorder='big')
    return out_byte
```

```python
def WriteByte(ser, bytein):
    out_byte = one_byte_int_to_serial_byte(bytein)
    ser.write(out_byte)
```

## Example

```python
byte1 = 7
WriteByte(ser,byte1)#<--
time.sleep(0.1)
byte2 = 156
WriteByte(ser,byte2)#<--
time.sleep(0.1)
next_line = read_one_line(ser)
extra = read_all(ser)
print('next_line: %s' % next_line)
print('extra: %s' % extra)
```
